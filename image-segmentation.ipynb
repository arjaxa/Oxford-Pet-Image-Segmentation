```python
import os
import tensorflow as tf

images_dir = os.path.join(dataset_dir, "images")
trimaps_dir = os.path.join(dataset_dir, "annotations", "trimaps")

IMAGE_SIZE = (128, 128)

with open(os.path.join(dataset_dir, "annotations", "trainval.txt"), "r") as f:
    trainval_names = [
        line.strip().split()[0]
        for line in f
        if line.strip()
    ]

with open(os.path.join(dataset_dir, "annotations", "test.txt"), "r") as f:
    test_names = [
        line.strip().split()[0]
        for line in f
        if line.strip()
    ]

def make_paths(names):
    image_paths = []
    mask_paths = []

    for name in names:
        image_path = os.path.join(images_dir, name + ".jpg")
        mask_path = os.path.join(trimaps_dir, name + ".png")

        if os.path.exists(image_path) and os.path.exists(mask_path):
            image_paths.append(image_path)
            mask_paths.append(mask_path)

    return image_paths, mask_paths

trainval_images, trainval_masks = make_paths(trainval_names)
test_images, test_masks = make_paths(test_names)

print("Train/validation images:", len(trainval_images))
print("Train/validation masks:", len(trainval_masks))
print("Test images:", len(test_images))
print("Test masks:", len(test_masks))
```

```python
import tensorflow as tf

def load_image_mask(image_path, mask_path):
    
    image = tf.io.read_file(image_path)
    image = tf.image.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, IMAGE_SIZE)
    image = tf.cast(image, tf.float32) / 255.0

    
    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=1)
    mask = tf.image.resize(
        mask,
        IMAGE_SIZE,
        method="nearest"
    )

    # converting labels from 1,2,3 → 0,1,2
    mask = tf.cast(mask, tf.int32) - 1

    return image, mask
```

```python
image, mask = load_image_mask(
    trainval_images[0],
    trainval_masks[0]
)

print("Image:", image.shape)
print("Image range:", tf.reduce_min(image).numpy(), "to", tf.reduce_max(image).numpy())

print("Mask:", mask.shape)
print("Mask classes:", tf.unique(tf.reshape(mask, [-1])).y.numpy())
```

```python
import random

pairs = list(zip(trainval_images, trainval_masks))

random.seed(42)
random.shuffle(pairs)

split_index = int(len(pairs) * 0.90)

train_pairs = pairs[:split_index]
val_pairs = pairs[split_index:]

train_images, train_masks = zip(*train_pairs)
val_images, val_masks = zip(*val_pairs)

train_images = list(train_images)
train_masks = list(train_masks)
val_images = list(val_images)
val_masks = list(val_masks)

print("Training:", len(train_images))
print("Validation:", len(val_images))
print("Test:", len(test_images))
```

```python
BATCH_SIZE = 16

def create_dataset(image_paths, mask_paths, shuffle=False):
    dataset = tf.data.Dataset.from_tensor_slices(
        (image_paths, mask_paths)
    )

    if shuffle:
        dataset = dataset.shuffle(
            buffer_size=len(image_paths),
            seed=42
        )

    dataset = dataset.map(
        load_image_mask,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset


train_dataset = create_dataset(
    train_images,
    train_masks,
    shuffle=True
)

val_dataset = create_dataset(
    val_images,
    val_masks
)

test_dataset = create_dataset(
    test_images,
    test_masks
)

print("datasets created.")
```

```python
from tensorflow.keras import layers, Model

inputs = layers.Input(shape=(128, 128, 3))

# encoder
x1 = layers.Conv2D(32, 3, padding="same", activation="relu")(inputs)
x1 = layers.Conv2D(32, 3, padding="same", activation="relu")(x1)
p1 = layers.MaxPooling2D()(x1)

x2 = layers.Conv2D(64, 3, padding="same", activation="relu")(p1)
x2 = layers.Conv2D(64, 3, padding="same", activation="relu")(x2)
p2 = layers.MaxPooling2D()(x2)

x3 = layers.Conv2D(128, 3, padding="same", activation="relu")(p2)
x3 = layers.Conv2D(128, 3, padding="same", activation="relu")(x3)
p3 = layers.MaxPooling2D()(x3)

# bottleneck
b = layers.Conv2D(256, 3, padding="same", activation="relu")(p3)
b = layers.Conv2D(256, 3, padding="same", activation="relu")(b)

# decoder
u3 = layers.UpSampling2D()(b)
u3 = layers.Concatenate()([u3, x3])
u3 = layers.Conv2D(128, 3, padding="same", activation="relu")(u3)
u3 = layers.Conv2D(128, 3, padding="same", activation="relu")(u3)

u2 = layers.UpSampling2D()(u3)
u2 = layers.Concatenate()([u2, x2])
u2 = layers.Conv2D(64, 3, padding="same", activation="relu")(u2)
u2 = layers.Conv2D(64, 3, padding="same", activation="relu")(u2)

u1 = layers.UpSampling2D()(u2)
u1 = layers.Concatenate()([u1, x1])
u1 = layers.Conv2D(32, 3, padding="same", activation="relu")(u1)
u1 = layers.Conv2D(32, 3, padding="same", activation="relu")(u1)

# pixel prediction
outputs = layers.Conv2D(
    3,
    1,
    padding="same",
    activation="softmax"
)(u1)

model = Model(inputs, outputs)

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=["accuracy"]
)

print("model compiled.")
```

```python
import numpy as np

total_intersection = 0
total_union = 0

for images, masks in test_dataset:
    predictions = model.predict(images, verbose=0)

    predicted_masks = np.argmax(predictions, axis=-1)
    true_masks = masks.numpy().squeeze(-1)

    true_foreground = true_masks != 1
    predicted_foreground = predicted_masks != 0

    intersection = np.logical_and(
        true_foreground,
        predicted_foreground
    ).sum()

    union = np.logical_or(
        true_foreground,
        predicted_foreground
    ).sum()

    total_intersection += intersection
    total_union += union

iou = total_intersection / total_union

print(f"Foreground IoU: {iou:.4f}")
```

```python
import numpy as np

num_classes = 3

intersection = np.zeros(num_classes)
union = np.zeros(num_classes)

for images, masks in test_dataset:
    predictions = model.predict(images, verbose=0)

    predicted_masks = np.argmax(predictions, axis=-1)
    true_masks = masks.numpy().squeeze(-1)

    for class_id in range(num_classes):
        true_class = (true_masks == class_id)
        predicted_class = (predicted_masks == class_id)

        intersection[class_id] += np.logical_and(
            true_class,
            predicted_class
        ).sum()

        union[class_id] += np.logical_or(
            true_class,
            predicted_class
        ).sum()

ious = intersection / (union + 1e-7)

for class_id, iou_value in enumerate(ious):
    print(f"Class {class_id} IoU: {iou_value:.4f}")

print(f"\nMean IoU: {np.mean(ious):.4f}")
```
